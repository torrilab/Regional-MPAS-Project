```python
#GETTING BOUNDING BOX centered at Houston, TX Mobile Facility (TRACER) Facility S2 ==> CSAP (C-Band Scanning ARM Precipitation Radar)
# 29°31'55"N, 95°17'2"W
longitude = coordinates.DMSToDecimal(95,17,2,hemisphere='W')
latitude = coordinates.DMSToDecimal(29,31,55,hemisphere='N')

dlon=coordinates.dxTOdlon(dx_m=250e3, lat_deg=latitude)
dlat=coordinates.dyTOdlat(dy_m=250e3)

N, W, S, E = [latitude + dlat, longitude - dlon,
              latitude - dlat, longitude + dlon]
print([N, W, S, E])
# Round outward to 0.25 grid
N = math.ceil(N / 0.25) * 0.25     # round north up
S = math.floor(S / 0.25) * 0.25    # round south down
W = math.floor(W / 0.25) * 0.25    # round west down (more negative)
E = math.ceil(E / 0.25) * 0.25     # round east up

print("==>",[N, W, S, E])
```

```python
#DOWNLOADING ERA5

#"Download ERA data with python" Code Inspired by https://github.com/joaohenry23/Download_ERA5_with_python
# https://cds.climate.copernicus.eu/how-to-api

# pip install cdsapi
# pip install "cdsapi>=0.7.4"
import cdsapi
from tqdm import tqdm
import os



c = cdsapi.Client()

variables = [
    'u_component_of_wind','v_component_of_wind',
    # 'vertical_velocity',
    'divergence','vorticity',
    'temperature',
    'specific_humidity','specific_cloud_liquid_water_content',
    'specific_cloud_ice_water_content','specific_rain_water_content',
    'relative_humidity','cloud_cover','geopotential'
]

for variable in tqdm(variables, desc="Downloading ERA5 variables"):
    
    c.retrieve(
        'reanalysis-era5-pressure-levels',
        {
            'product_type': 'reanalysis',
            'format': 'netcdf',
            'variable': 'vertical_velocity',
            'pressure_level': [
                '100', '250', '500', '750', '1000'
            ],
            'year': '2022',
            'month': ['06','07'],
            'day': ['29','30','01'],
            'time': [f"{h:02d}:00" for h in range(24)],
            'area': [N, W, S, E],
            'grid': [0.25, 0.25],
        },
        os.path.join(dataDirectory,f'{variable}_ERA5_2022-06-29_to_07-01.nc') 
    )
```
